```python
documents = [
    "The minimum attendance requirement for students is 85 percent.",
    "Students must pay examination fees before the announced deadline.",
    "The college library closes at 8 PM on working days.",
    "Internal assessment marks are published on the student portal.",
    "Students can contact the department office for timetable questions.",
    "The computer laboratory is available during scheduled practical sessions."
]

for i, document in enumerate(documents, start=1):
    print(f"{i}. {document}")
```

```python
from sentence_transformers import SentenceTransformer

for i, document in enumerate(documents, start=1):
    print(f"{i}. {document}")

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
print("Embedding model loaded successfully!")

document_embeddinga = embedding_model.encode(
    documents,
    convert_to_numpy=True
)

print("Number of dcouments:", len(documents))
print("Embedding matrix shape:", document_embeddinga.shape)
print("First vector:", document_embeddinga[0])
```

```python
query = "What is the required attendance percentage?"

query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True
)

print("Query:", query)
print("Embedding matrix shape:", query_embedding.shape)
```

```python
import faiss

dimension = document_embeddinga.shape[1]
faiss_index = faiss.IndexFlatL2(dimension)
faiss_index.add(document_embeddinga.astype("float32" ))

print("FAISS index created.")
print("Vectors stored",faiss_index.ntotal)
```

```python
import chromadb

chroma_client = chromadb.Client()
collection = chroma_client.get_or_create_collection(
    name="college_docs"
)
print("ChromaDB collection created.")
```

```python
document_ids = [f"doc{i}" for i in range(1, len(documents) + 1)]

collection.add(
    ids = document_ids,
    documents = documents,
    embeddings = document_embeddinga.tolist()
)

print("Documents added.")
print("Stored records:", collection.count())
```

```python
query = "What is the attendance requirement?"

query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True
)[0].tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=2)

print("Query:", query)
print("\nRetrieved documents:")

for rank, document in enumerate(results["documents"][0], start=1):
  print(f"{rank}. {document}")
```

```python
metadata_collection = chroma_client.get_or_create_collection(
    name = "college_docs_with_metadata"
)

metadata_values = [
    {"category": "attendance"},
    {"category": "fees"},
    {"category": "library"},
    {"category": "assessment"},
    {"category": "administration"},
    {"category": "laboratory"}
]

metadata_collection.add(
    ids = document_ids,
    documents = documents,
    embeddings = document_embeddinga.tolist(),
    metadatas = metadata_values
)

print("Documents added with metadata.")
print("Stored records:", metadata_collection.count())
```

```python
attendance_query = "What is the attendance requirement?"
attendance_vector = embedding_model.encode(
    [attendance_query],
    convert_to_numpy=True
)[0].tolist()

attendance_results = metadata_collection.query(
    query_embeddings = [attendance_vector],
    n_results = 2,
    where = {"category": "attendance"}
)

print("QUERY:", attendance_query)

for document in attendance_results["documents"][0]:
    print("Document:", document)
```

```python
def semantic_search_faiss(query, top_k=3):
    query_vector = embedding_model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = faiss_index.search(
        query_vector,
        top_k
    )

    results = []

    for distance, index in zip(distances[0], indices[0]):
        results.append({
            "document": documents[index],
            "distance": float(distance),
            "index": int(index)
        })

    return results


results = semantic_search_faiss(
    "Where can I find my internal marks?",
    top_k=3
)

for item in results:
    print(item)
```
